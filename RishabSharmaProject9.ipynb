```r
# code here
library(data.table)




deathDF <- fread("/anvil/projects/tdm/data/death_records/DeathRecords.csv", select = "Race")

read.csv("/anvil/projects/tdm/data/death_records/Race.csv")

sort(table(deathDF))

save_df <- subset(deathDF, Race != 1 & Race !=2 )

race_counts <- table(save_df$Race)



dotchart(sort(race_counts),
    ylab = "Differing races excluding White and Black",
    xlab = "Amount of deaths",
    main = "Amount of deaths per race excluding white and black")
```

I made my dotchart plot unique, as I sorted them by the amount of deaths per race in increasing order, so that you can  easily compare the amount of deaths per race. I also included a labeled x and Y axis, along with a title so the graph's information is clear.

I find dotchart way more insightful in terms of representing deaths per race, and that is because it displays the data in a much more digestible format, rather than the the long bars of the plot which in my opinion, make it harder to compare the smaller values against each other, even when the values are sorted. My final decision was to exclude the values 1 and 2 which correspond to the white and black deaths. I excluded these because they had a much higher amount of deaths as compared to the other races, that it made comparing the other races nearly impossible, as their values were much smaller and hard to see.

```r
# code here
library(data.table)
deathDF <- fread("/anvil/projects/tdm/data/death_records/DeathRecords.csv")
cleanDF <- subset(deathDF, MaritalStatus !="S" & MaritalStatus!="U" & MaritalStatus!="D" & Age!=999  )
boxplot(Age ~ MaritalStatus, data = cleanDF)

sixtyDF <- subset(deathDF, MaritalStatus !="S" & MaritalStatus!="U" & MaritalStatus!="D" & Age == 60  )

barplot(
    table(sixtyDF$MaritalStatus) ,
    xlab="Marital Status",
    ylab = "Count",
    main ="Marital Status of 60-Year-Olds")

eightyDF <- subset(deathDF, MaritalStatus !="S" & MaritalStatus!="U" & MaritalStatus!="D" & Age == 80  )

barplot(
    table(eightyDF$MaritalStatus) ,
    xlab="Marital Status",
    ylab = "Count",
    main ="Marital Status of 80-Year-Olds")

seventyDF <- subset(deathDF,  Age == 70  )

barplot(
    table(seventyDF$MaritalStatus) ,
    xlab="Marital Status",
    ylab = "Count",
    main ="Relationship Status of deceased 70-Year-Olds")

eightzeroDF <- subset(deathDF,  Age == 80  )

barplot(
    table(eightzeroDF$MaritalStatus) ,
    xlab="Marital Status",
    ylab = "Count",
    main ="Relationship Status of deceased 80-Year-Olds")

sixzeroDF <- subset(deathDF,  Age == 60  )

barplot(
    table(sixzeroDF$MaritalStatus) ,
    xlab="Marital Status",
    ylab = "Count",
    main ="Relationship Status of deceased 60-Year-Olds")
```
